```python
%load_ext autoreload
%autoreload 2
from pathlib import Path
import sys
ROOT = Path.cwd().resolve().parent 
sys.path.append(str(ROOT))
```

```python
from src.preprocessing import load_data, map_id_to_idx, build_restaurant_history_reviews, build_user_history_reviews, utility_mappings, val_test_split

DATA_PATH = ROOT/"data/filter_all_t.json"
data = load_data(path=DATA_PATH)
train_df, val_df, test_df = val_test_split(data)

print(len(train_df))
print(len(val_df))
print(len(test_df))
```

```python
user_id_to_idx, item_id_to_idx = utility_mappings(train_df)

train_df = map_id_to_idx(train_df, user_id_to_idx, item_id_to_idx)
val_df = map_id_to_idx(val_df, user_id_to_idx, item_id_to_idx)
test_df = map_id_to_idx(test_df, user_id_to_idx, item_id_to_idx)

user_profile_text = build_user_history_reviews(train_df)
restaurant_profile_text = build_restaurant_history_reviews(train_df)
```

```python
from libreco.data import DatasetPure
from libreco.algorithms import NCF
from libreco.evaluation import evaluate

train_cf = train_df[["user_id", "business_id", "rating"]].rename(
    columns={
        "user_id": "user",
        "business_id": "item",
        "rating": "label"
    }
)

val_cf = val_df[["user_id", "business_id", "rating"]].rename(
    columns={
        "user_id": "user",
        "business_id": "item",
        "rating": "label"
    }
)

train_data, data_info = DatasetPure.build_trainset(train_cf)
eval_data = DatasetPure.build_evalset(val_cf)

ncf = NCF(
    task="rating",
    data_info=data_info,
    loss_type="rmse",
    embed_size=60,
    n_epochs=15,
    lr=1e-3,
    batch_size=256,
)

ncf.fit(
    train_data,
    eval_data=eval_data,
    neg_sampling=False,
    metrics=["rmse"]
)
```

```python
test_cf = test_df[["user_id", "business_id", "rating"]].rename(
    columns={
        "user_id": "user",
        "business_id": "item",
        "rating": "label"
    }
)

test_data = DatasetPure.build_testset(test_cf)

result = evaluate(
    model=ncf,
    data=test_cf,
    neg_sampling=False,
    metrics=["rmse"]
)

print(result)
```
